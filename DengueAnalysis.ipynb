## 1) Importação e Tratamento dos Dados

**Detalhamento dos Dados e Abordagem de Extração**

A tabela gerada a partir da consulta SQL contém informações essenciais para a análise epidemiológica da dengue no Brasil. Cada registro na tabela representa um agrupamento de notificações, estruturado com as seguintes colunas:

- **ano**: Ano em que a notificação foi registrada.
- **semana_notificacao**: Semana epidemiológica correspondente à data de notificação.
- **data_notificacao**: Data exata da notificação do caso.
- **sigla_uf_notificacao**: Unidade Federativa (UF) onde o caso foi notificado, identificada pela sigla.
- **id_municipio_notificacao**: Código IBGE do município onde o caso foi notificado.
- **total_notificacoes**: Contagem total de notificações para o agrupamento especificado.

#### Estratégia de Extração e Transformação

Os dados foram extraídos diretamente da base pública `basedosdados.br_ms_sinan.microdados_dengue` utilizando uma consulta SQL otimizada. A query realiza as seguintes operações:

1. **Extração de Semana Epidemiológica**: A função `EXTRACT(WEEK FROM DATE(data_notificacao))` foi utilizada para derivar a semana epidemiológica a partir da data de notificação, permitindo análises temporais mais detalhadas.
2. **Agrupamento**: Os dados foram agrupados por ano, data de notificação, UF e município, reduzindo a granularidade dos dados originais (nível individual) para um formato mais adequado à análise de tendências.
3. **Contagem de Notificações**: A contagem total de notificações foi calculada para cada agrupamento, armazenada na coluna `total_notificacoes`.
4. **Ordenação**: Os resultados foram ordenados por ano, UF e município, garantindo uma estrutura lógica e consistente para análises subsequentes.

```python
# Consulta SQL
query = """
SELECT
  EXTRACT(YEAR FROM DATE(data_notificacao)) AS ano,
  EXTRACT(WEEK FROM DATE(data_notificacao)) AS semana_notificacao,
  DATE(data_notificacao) AS data_notificacao,
  CASE SUBSTR(id_municipio_notificacao, 1, 2)
    WHEN '11' THEN 'RO'
    WHEN '12' THEN 'AC'
    WHEN '13' THEN 'AM'
    WHEN '14' THEN 'RR'
    WHEN '15' THEN 'PA'
    WHEN '16' THEN 'AP'
    WHEN '17' THEN 'TO'
    WHEN '21' THEN 'MA'
    WHEN '22' THEN 'PI'
    WHEN '23' THEN 'CE'
    WHEN '24' THEN 'RN'
    WHEN '25' THEN 'PB'
    WHEN '26' THEN 'PE'
    WHEN '27' THEN 'AL'
    WHEN '28' THEN 'SE'
    WHEN '29' THEN 'BA'
    WHEN '31' THEN 'MG'
    WHEN '32' THEN 'ES'
    WHEN '33' THEN 'RJ'
    WHEN '35' THEN 'SP'
    WHEN '41' THEN 'PR'
    WHEN '42' THEN 'SC'
    WHEN '43' THEN 'RS'
    WHEN '50' THEN 'MS'
    WHEN '51' THEN 'MT'
    WHEN '52' THEN 'GO'
    WHEN '53' THEN 'DF'
    ELSE NULL
  END AS sigla_uf_notificacao,
  id_municipio_notificacao,
  COUNT(*) AS total_notificacoes
FROM
  `basedosdados`.`br_ms_sinan`.`microdados_dengue`
GROUP BY
  1, 2, 3, 4, 5
ORDER BY
  data_notificacao,
  sigla_uf_notificacao,
  id_municipio_notificacao;
"""
df = client.query(query).to_dataframe()

# Dicionário de regiões por estado
regioes_br = {
    'AC': 'Norte', 'AP': 'Norte', 'AM': 'Norte', 'PA': 'Norte', 'RO': 'Norte', 'RR': 'Norte', 'TO': 'Norte',
    'AL': 'Nordeste', 'BA': 'Nordeste', 'CE': 'Nordeste', 'MA': 'Nordeste', 'PB': 'Nordeste',
    'PE': 'Nordeste', 'PI': 'Nordeste', 'RN': 'Nordeste', 'SE': 'Nordeste',
    'DF': 'Centro-Oeste', 'GO': 'Centro-Oeste', 'MT': 'Centro-Oeste', 'MS': 'Centro-Oeste',
    'ES': 'Sudeste', 'MG': 'Sudeste', 'RJ': 'Sudeste', 'SP': 'Sudeste',
    'PR': 'Sul', 'RS': 'Sul', 'SC': 'Sul'
}

# Gerando coluna com o depara de Região do Brasil
df["regiao"] = df["sigla_uf_notificacao"].map(regioes_br)

# Remover linhas com valores nulos na coluna 'data_notificacao'
df = df.dropna(subset=['data_notificacao'])
```

### Mapeamento Geoespacial

A abordagem de visualização geoespacial é extremamente útil e prática para se estudar/ compreender o movimento de uma epidemia.

Com os gráficos abaixo, será possível identificar um fenômeno de "migração" do vírus para regiões conhecidas por seu clima mais frio e seco.

Vemos uma tendência do vírus a ir de encontro ao Sul e Sudeste nos últimos 4/5 anos.

```python
# Carregar dados dos municípios
url = "https://raw.githubusercontent.com/kelvins/municipios-brasileiros/refs/heads/main/csv/municipios.csv"
df_municipios = pd.read_csv(url)

# Preparar para a união de tabelas
df_municipios["codigo_ibge"] = df_municipios["codigo_ibge"].astype(str).str.zfill(7)
df_municipios = df_municipios.rename(columns={
    "codigo_ibge": "id_municipio_notificacao",
    "nome": "nome_municipio",
    "uf": "sigla_uf_municipio"
})

# Merge com o df principal
df_merged = pd.merge(
    df,
    df_municipios[["id_municipio_notificacao", "nome_municipio", "latitude", "longitude"]],
    on="id_municipio_notificacao",
    how="left"
)

# Gerando série temporal mensal
df_merged['data_notificacao'] = pd.to_datetime(df_merged['data_notificacao'])
df_merged['ano_mes'] = df_merged['data_notificacao'].dt.to_period('M').astype(str)

# Criando df's com níveis de hierarquia temporal diferentes, caso haja necessidade posterior em avaliar a movimentação do vírus de forma mensal/anual
df_merged_mes = df_merged.groupby(['ano_mes', 'id_municipio_notificacao', 'latitude', 'longitude'], as_index=False)['total_notificacoes'].sum()
df_merged_ano = df_merged.groupby(['ano', 'id_municipio_notificacao', 'latitude', 'longitude'], as_index=False)['total_notificacoes'].sum()
```

Gerando mapa de notificações por ano por município

Função para estruturar uma série temporal completa por município.

Muitas séries têm datas faltantes (dias sem notificação), mas para modelos de séries temporais isso precisa ser preenchido corretamente (com `0` ou interpolado).

* **Interpolação linear (mais elegante e contínua)**
Mantém a fluidez da série.
Útil quando o padrão de notificações é relativamente contínuo (sem picos abruptos).
Preserva a estrutura estatística da série para modelos como ARIMA/SARIMA.

* **Preenchimento com zero (assume ausência de casos)**
Reforça que ausência de registro é ausência de evento.
Pode distorcer séries em regiões com subnotificação.
Pode enviesar modelos que interpretam longos períodos de “calmaria”.

* **Interpolação por média móvel (mais suave)**
Considera a média de dias anteriores/seguintes.
Boa para manter tendência sem saltos artificiais.



```python
def preencher_serie(serie, metodo='interpolacao'):
    """
    Preenche os valores ausentes de uma série temporal.
    Opções:
        - 'interpolacao': usa interpolação linear
        - 'zeros': preenche com zero
        - 'media_movel': usa média móvel centrada + interpolação complementar
    """
    if metodo == 'interpolacao':
        return serie.interpolate(method='linear')
    elif metodo == 'zeros':
        return serie.fillna(0)
    elif metodo == 'media_movel':
        serie_media = serie.rolling(window=3, center=True).mean()
        serie_final = serie.fillna(serie_media)
        return serie_final.interpolate(method='linear')
    else:
        raise ValueError("Método inválido. Escolha: 'interpolacao', 'zeros' ou 'media_movel'.")
```

```python
# Parâmetro de preenchimento (você pode trocar para 'interpolacao' ou 'zeros' se quiser testar depois)
metodo_preenchimento = 'media_movel'
```

```python
# Dicionário para armazenar as séries preenchidas por município
series_municipios = {}

# Loop sobre todos os municípios
for municipio in df_grouped['nome_municipio'].unique():

    # Filtrar dados do município
    df_mun = df_grouped[df_grouped['nome_municipio'] == municipio].copy()

    # Criar série com índice datetime completo
    full_range = pd.date_range(start=df_mun['date'].min(), end=df_mun['date'].max(), freq='D')
    df_mun = df_mun.set_index('date').reindex(full_range).rename_axis('date').reset_index()
    df_mun['nome_municipio'] = municipio

    # Preencher total_notificacoes com NaN inicialmente
    df_mun['total_notificacoes'] = df_mun['total_notificacoes'].fillna(np.nan)

    # Aplicar o método escolhido (média móvel no seu caso)
    df_mun['total_notificacoes'] = preencher_serie(df_mun['total_notificacoes'], metodo=metodo_preenchimento)

    # Armazenar no dicionário
    series_municipios[municipio] = df_mun
```

##### Testes Estatísticos iniciais

Antes de aplicar qualquer modelo preditivo de séries temporais, precisamos entender a natureza da série. Esses testes ajudam a:

* Verificar estacionariedade (condição essencial para ARIMA/SARIMA)
* Detectar autocorrelação ou sazonalidade
* Avaliar se transformações são necessárias (ex: diferenciação)

A **estacionariedade** em séries temporais refere-se à propriedade de as estatísticas de uma série (como a média, a variância e a autocovariância) não mudarem com o tempo. Em outras palavras, a forma como a série está a flutuar permanece a mesma, independentemente de quando é observada.

"Toda série estacionária pode ser representada por um modelo ARMA (p,q) com baixos valores de `p` e `q`. Este modelo é suficiente para estimar uma série temporal." *Teorema de Wald*

**ADF - Augmented Dickey-Fuller (mais usado)**

- O que testa:

Estacionariedade (se a média e variância da série são constantes ao longo do tempo)

- Hipóteses:

H₀ (nula): a série não é estacionária

H₁ (alternativa): a série é estacionária

- Como interpretar:

p-valor < 0.05 → rejeita H₀ → série estacionária ✅

p-valor > 0.05 → não rejeita H₀ → série não estacionária ❌

**KPSS - Kwiatkowski–Phillips–Schmidt–Shin (*complementar ao ADF*)**

- O que testa:

Estacionariedade, mas com hipótese oposta

- Hipóteses:

H₀: a série é estacionária

H₁: a série não é estacionária

- Quando usar:

Em conjunto com ADF → abordagem de confirmação cruzada

*ADF e KPSS juntos funcionam como “testes espelho”.*

**PP - Phillips-Perron (*complementar ao ADF*)**

- O que testa:

Estacionariedade, mas com hipótese oposta

- Hipóteses:

H₀: a série não é estacionária

H₁: a série é estacionária

```python
# Função para teste ADF
def testar_adf(serie, nome_municipio):
    resultado = adfuller(serie.dropna(), autolag='AIC')
    return {
        'municipio': nome_municipio,
        'ADF_statistic': resultado[0],
        'ADF_p_value': resultado[1],
        'ADF_stationary': resultado[1] < 0.05
    }

# Função para teste KPSS
def testar_kpss(serie, nome_municipio, regression='c'):
    resultado = kpss(serie.dropna(), regression=regression, nlags='auto')
    return {
        'municipio': nome_municipio,
        'KPSS_statistic': resultado[0],
        'KPSS_p_value': resultado[1],
        'KPSS_stationary': resultado[1] > 0.05
    }

# Função para teste PP
def testar_pp(serie, nome_municipio):
    resultado = PhillipsPerron(serie.dropna()).stat
    p_valor = PhillipsPerron(serie.dropna()).pvalue
    return {
        'municipio': nome_municipio,
        'PP_statistic': resultado,
        'PP_p_value': p_valor,
        'PP_stationary': p_valor < 0.05
    }

# Diagnóstico final combinando ADF e KPSS
def diagnostico_final(adf_stationary, kpss_stationary, pp_stationary):
    votos = [adf_stationary, kpss_stationary, pp_stationary]
    if votos.count(True) >= 2:
        return 'Estacionária'
    elif votos.count(False) >= 2:
        return 'Não estacionária'
    else:
        return 'Inconclusiva'
```

```python
# Lista para armazenar os resultados
resultados_estatisticos = []

# Loop por município usando o dicionário de séries preenchidas
for municipio, df_mun in series_municipios.items():
    serie = df_mun['total_notificacoes']

    # Aplicar os três testes
    adf = testar_adf(serie, municipio)
    kpss_ = testar_kpss(serie, municipio)
    pp = testar_pp(serie, municipio)

    # Diagnóstico combinando os três testes
    status_final = diagnostico_final(adf['ADF_stationary'], kpss_['KPSS_stationary'], pp['PP_stationary'])

    # Armazenar resultados
    resultados_estatisticos.append({
        'Município': municipio,
        'ADF Estatística': round(adf['ADF_statistic'], 4),
        'ADF p-valor': round(adf['ADF_p_value'], 4),
        'ADF Estacionária': adf['ADF_stationary'],
        'KPSS Estatística': round(kpss_['KPSS_statistic'], 4),
        'KPSS p-valor': round(kpss_['KPSS_p_value'], 4),
        'KPSS Estacionária': kpss_['KPSS_stationary'],
        'PP Estatística': round(pp['PP_statistic'], 4),
        'PP p-valor': round(pp['PP_p_value'], 4),
        'PP Estacionária': pp['PP_stationary'],
        'Diagnóstico Final': status_final
    })

# Criar DataFrame final
df_resultados = pd.DataFrame(resultados_estatisticos)

# Exibir a tabela
df_resultados
```

##### Modelagem Preditiva com Modelos Simples de Séries Temporais


Nesta etapa, iniciamos a aplicação de **modelos preditivos univariados simples**, ou seja, modelos que utilizam apenas o histórico da própria série de casos de dengue para realizar previsões. A proposta é estabelecer uma **linha de base (baseline)** para comparação futura com modelos mais sofisticados (como ARIMA, SARIMA, e modelos multivariados que incluam clima e fatores socioeconômicos).

**Modelos Aplicados**

Aplicaremos os seguintes modelos simples:

- ***Naive Forecast*** (Modelo Persistente): assume que o último valor observado continuará igual nos períodos seguintes.
- ***Mean Forecast*** (Média): prevê o valor futuro com base na média de todos os valores passados.
- ***Drift Forecast*** (Tendência Linear): projeta uma tendência linear entre o primeiro e o último ponto do treino.

Esses modelos são fundamentais por sua **simplicidade interpretativa**, **rapidez de execução** e por servirem como **referência mínima aceitável** para outros modelos mais complexos.

---

**Métricas de Avaliação**

Para cada município e modelo, calcularemos as seguintes métricas de erro:

- ***MAE (Mean Absolute Error)***: erro médio absoluto, fácil de interpretar.
- ***RMSE (Root Mean Squared Error)***: penaliza mais os grandes erros.
- ***MAPE (Mean Absolute Percentage Error)***: erro percentual médio (evitar se a série contiver muitos valores próximos de 0).

Essas métricas serão armazenadas em uma tabela comparativa, permitindo identificar **qual modelo teve o melhor desempenho em cada município**.

Isso reforça a importância de tratar os modelos simples como **baseline**, e não como solução definitiva.*

```python
# Função para converter a série numa série semanal e não diária
def preparar_serie_semanal(df_mun, data_fim_treino='2022-12-31', data_fim_teste='2023-12-31'):
    """
    Agrupa série por semana (frequência semanal), e separa treino e teste com base nas datas fornecidas.
    """
    df_mun['date'] = pd.to_datetime(df_mun['date'])
    df_mun = df_mun.set_index('date')

    # Agregar semanalmente (domingo é o fechamento da semana)
    serie_semanal = df_mun['total_notificacoes'].resample('W').sum()

    # Separar treino e teste
    treino = serie_semanal[:data_fim_treino]
    teste = serie_semanal[data_fim_treino: data_fim_teste]

    return treino, teste, serie_semanal
```

```python
# Gerando as funções dos modelos para receberem a série de treino e a quantidade de períodos de teste, e retornarem previsões.

# Modelo Naive (último valor)
def naive_forecast(treino, h):
    return np.repeat(treino.iloc[-1], h)

# Modelo Mean Forecast (média histórica)
def mean_forecast(treino, h):
    return np.repeat(treino.mean(), h)

# Modelo Drift Forecast (tendência linear)
def drift_forecast(treino, h):
    y0 = treino.iloc[0]
    yT = treino.iloc[-1]
    return [y0 + (i + 1) * (yT - y0) / (len(treino) - 1) for i in range(h)]
```

```python
# Calcula as três métricas de erro

def calcular_metricas(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)  # cálculo direto do RMSE
    mape = np.mean(np.abs((y_true - y_pred) / np.where(y_true == 0, 1, y_true))) * 100  # evitar divisão por zero
    return mae, rmse, mape
```

```python
# Loop por município com separação de treino/teste

resultados_modelos = []

for municipio, df_mun in series_municipios.items():
    # Preparar série semanal e separar treino/teste
    treino, teste, _ = preparar_serie_semanal(df_mun, data_fim_treino='2022-12-31', data_fim_teste='2023-12-31')
    h = len(teste)

    if h == 0:
        continue  # pular cidades sem dados suficientes

    # Previsões
    previsoes = {
        'Naive': naive_forecast(treino, h),
        'Mean': mean_forecast(treino, h),
        'Drift': drift_forecast(treino, h)
    }

    for modelo_nome, pred in previsoes.items():
        mae, rmse, mape = calcular_metricas(teste.values, pred)
        resultados_modelos.append({
            'Município': municipio,
            'Modelo': modelo_nome,
            'MAE': round(mae, 2),
            'RMSE': round(rmse, 2),
            'MAPE': round(mape, 2)
        })
```

##### Modelos Clássicos de Suavização Exponencial

**Modelagem com Suavização Exponencial**

Nesta etapa, avançamos na complexidade dos modelos preditivos univariados, aplicando técnicas clássicas de **suavização exponencial**. Esses modelos são bastante utilizados para séries temporais, especialmente quando há **padrões de tendência ou sazonalidade** que os modelos simples (Naive, Média, Drift) não capturam bem.

A suavização exponencial dá **mais peso aos valores mais recentes da série**, reduzindo o impacto de variações antigas e tornando a previsão mais responsiva ao comportamento atual.

---

**Modelos Aplicados:**

| Modelo                         | Finalidade                                               |
|--------------------------------|-----------------------------------------------------------|
| **SES (Simple Exponential Smoothing)** | Ideal para séries **sem tendência nem sazonalidade**. Trabalha apenas com o **nível** da série. |
| **Holt (Linear)**              | Útil para séries com **tendência**, mas **sem sazonalidade**. Trabalha com **nível + tendência**. |
| **Holt-Winters Aditivo**       | Para séries com **sazonalidade constante** (ex: flutuações fixas ao longo do tempo). Trabalha com **nível + tendência + sazonalidade aditiva**. |
| **Holt-Winters Multiplicativo**| Para séries com **sazonalidade proporcional** (ex: sazonalidade cresce junto com a tendência). Trabalha com **nível + tendência + sazonalidade multiplicativa**. |

*Obs: O modelo Holt-Winters multiplicativo requer que os valores da série sejam estritamente positivos e diferentes de zero. Isso ocorre porque ele aplica multiplicações com os componentes sazonais, e multiplicar por zero ou por valores negativos quebra o modelo. Séries de notificação de doenças geralmente apresentam sazonalidade aditiva — picos ocorrem em certas épocas do ano, mas o tamanho do pico não necessariamente cresce proporcionalmente ao volume base da série.*


**Avaliação dos Modelos**

Para cada modelo e município, avaliaremos o desempenho com base nas seguintes métricas:

- **MAE (Mean Absolute Error)**: erro médio absoluto
- **RMSE (Root Mean Squared Error)**: penaliza mais os grandes erros
- **MAPE (Mean Absolute Percentage Error)**: erro percentual médio (menos confiável quando há muitos zeros)

Os resultados serão organizados em tabelas comparativas para identificar **qual modelo obteve melhor desempenho em cada município**.

##### Modelos ARIMA e SARIMA

Os modelos **ARIMA (AutoRegressive Integrated Moving Average)** e sua versão sazonal, **SARIMA (Seasonal ARIMA)**, são amplamente utilizados para séries temporais que apresentam:

- Tendência;
- Autocorrelação;
- Sazonalidade (no caso do SARIMA).

A ideia é modelar explicitamente os padrões de **defasagem (lags)** e **diferença** entre os valores da série, permitindo capturar tanto o comportamento passado quanto oscilações sazonais.

**Notação ARIMA(p, d, q) – Modelo não sazonal**

AR (AutoRegressive) → parte p

Modela a dependência de um valor com seus valores anteriores (lags).
Ex: número de casos esta semana depende das últimas 3 semanas.

I (Integrated) → parte d

Número de diferenciações para tornar a série estacionária.
Ex: se a série tem tendência de subida, podemos calcular as diferenças entre os valores (delta), como x[t] - x[t-1].

MA (Moving Average) → parte q

Usa erros de previsão anteriores como parte do modelo.
Ex: corrige a previsão com base nos erros cometidos antes.

- **p:** número de defasagens autorregressivas (AR)
- **d:** número de diferenciações para tornar a série estacionária
- **q:** número de termos de média móvel (MA)

**Notação SARIMA(p, d, q)(P, D, Q, s)**
- Adiciona os mesmos conceitos de AR, I, MA **para os componentes sazonais**
- **s:** periodicidade sazonal (ex: `s = 52` para dados semanais)

###### Modelos ARIMA

```python
# Funções auxiliares
def calcular_metricas(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((y_true - y_pred) / np.maximum(np.abs(y_true), 1))) * 100
    return mae, rmse, mape
```

```python
# ARIMA

# Lista para armazenar os resultados
resultados_gerais = []

# Parâmetros do ARIMA
p_values = range(0, 4)
d_values = range(0, 2)
q_values = range(0, 4)
```

```python
# Loop por município
for municipio, df_mun in tqdm(series_municipios.items()):
    try:
        treino, teste, serie_completa = preparar_serie_semanal(df_mun)
        h = len(teste)
        if h == 0:
            continue

        melhor_modelos = []

        for p in p_values:
            for d in d_values:
                for q in q_values:
                    try:
                        modelo = ARIMA(treino, order=(p, d, q)).fit()
                        pred = modelo.forecast(steps=h)

                        mae, rmse, mape = calcular_metricas(teste.values, pred)

                        melhor_modelos.append({
                            'Município': municipio,
                            'Modelo': f'ARIMA({p},{d},{q})',
                            'MAE': mae,
                            'RMSE': rmse,
                            'MAPE': mape,
                            'Previsão': pred,
                            'Real': teste
                        })
                    except:
                        continue

        # Selecionar os top 3 modelos por menor RMSE
        top3 = sorted(melhor_modelos, key=lambda x: x['RMSE'])[:3]
        resultados_gerais.extend(top3)

    except Exception as e:
        print(f"Erro ao processar {municipio}: {e}")
```

###### Modelos SARIMA

Antes de realizarmos as previsões com o viés sazonal que o SARIMA traz, vamos revisar os termos do modelo:

| Termo | Nome                     | Significado prático                                                                |
| ----- | ------------------------ | ---------------------------------------------------------------------------------- |
| `p`   | AutoRegressive (AR)      | Quantidade de defasagens passadas da **própria série** que são usadas para prever. |
| `d`   | Diferença (Differencing) | Número de vezes que a série será diferenciada para torná-la **estacionária**.      |
| `q`   | Moving Average (MA)      | Quantidade de defasagens passadas de **erros/resíduos** usados na previsão.        |



---


Agora o que realmente diferencia o SARIMA do ARIMA, são os seguintes termos:

| Termo | Nome                           | Significado prático                                                                                          |
| ----- | ------------------------------ | ------------------------------------------------------------------------------------------------------------ |
| `P`   | AR Sazonal (SAR)               | Quantas **defasagens sazonais** da série são usadas (ex: 1 ano atrás, se `s=52`).                            |
| `D`   | Diferença sazonal (Seasonal D) | Quantas vezes aplicar **diferença entre sazonas** (ex: valor desta semana vs. mesma semana no ano anterior). |
| `Q`   | MA Sazonal (SMA)               | Quantas **defasagens sazonais de erro** são usadas.                                                          |
| `s`   | Período Sazonal                | Tamanho do ciclo sazonal. Para dados semanais com padrão anual: `s=52`.                                      |


Em termos mais simples:

`p` e `q`: capturam o comportamento no curto prazo (dias/semanas atrás).

`P` e `Q`: capturam repetições sazonais (ex: semanas de março sempre têm pico?).

`d` e `D`: corrigem tendências (geral) e sazonalidade estrutural.

```python
# SARIMA

# Função para calcular métricas
def calcular_metricas(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((y_true - y_pred) / np.maximum(np.abs(y_true), 1))) * 100
    return mae, rmse, mape

# Função de Grid Search SARIMA
# Lista dos modelos reduzidos e inteligentes
modelos_sarima_sugeridos = [
    {'order': (1, 1, 1), 'seasonal_order': (1, 1, 1)},
    {'order': (2, 1, 2), 'seasonal_order': (1, 1, 1)},
    {'order': (1, 1, 1), 'seasonal_order': (0, 1, 1)},
    {'order': (1, 1, 0), 'seasonal_order': (1, 1, 1)},
    {'order': (2, 1, 0), 'seasonal_order': (1, 1, 0)},
    {'order': (0, 1, 1), 'seasonal_order': (0, 1, 1)},
    {'order': (1, 1, 2), 'seasonal_order': (1, 1, 0)},
]

# Função
def grid_search_sarima_custom(treino, teste, s=52):
    resultados = []
    h = len(teste)

    for cfg in modelos_sarima_sugeridos:
        try:
            modelo = SARIMAX(
                treino,
                order=cfg['order'],
                seasonal_order=cfg['seasonal_order'] + (s,),
                enforce_stationarity=False,
                enforce_invertibility=False
            ).fit(disp=False)

            previsao = modelo.forecast(steps=h)
            mae, rmse, mape = calcular_metricas(teste.values, previsao)

            resultados.append({
                'order': cfg['order'],
                'seasonal_order': cfg['seasonal_order'],
                'MAE': mae,
                'RMSE': rmse,
                'MAPE': mape,
                'forecast': previsao,
                'real': teste
            })
        except Exception as e:
            print(f"Erro SARIMA{cfg['order']}{cfg['seasonal_order'] + (s,)}: {e}")
            continue

    return sorted(resultados, key=lambda x: x['RMSE'])[:3]
```

Com a lista de estações em mãos, precisamos encontrar qual estação está mais próxima de cada município presente em `df_sudeste`.

Usaremos as coordenadas geográficas para isso. Uma abordagem eficiente é utilizar um algoritmo de vizinho mais próximo (*k-d tree* ou *Ball Tree*) para encontrar, para cada ponto (município), o ponto de estação mais próximo.

Preparação dos dados:
* Certifique-se de que o `df_sudeste` tenha colunas de latitude e longitude numéricas.
* Converteremos as latitudes/longitudes para radianos, já que utilizaremos a ***métrica de Haversine*** para distâncias geográficas esféricas.

A ***métrica de Haversine*** é uma fórmula que calcula a distância ao longo da superfície de uma esfera (*como a Terra*) entre dois pontos, dados em coordenadas de latitude e longitude.

Essa fórmula é amplamente usada em navegação e sistemas de mapas para medir distâncias “em linha reta” (sobre a superfície terrestre) entre cidades, pontos de GPS, estações, etc., com boa precisão.

Essa fórmula resulta no comprimento do arco do círculo máximo entre os pontos – equivalente à menor distância sobre a superfície esférica que conecta os dois pontos.

Como mencionado, a Terra não é uma esfera perfeita (é um esferoide oblato), mas a diferença típica ao usar Haversine (esfera) em vez de um modelo elipsoidal mais preciso é muito pequena, o que é aceitável para a maioria das aplicações práticas.

No código, usamos ***Ball Tree***  com métrica Haversine para encontrar o índice da estação mais próxima (`nearest_station_index`) para cada coordenada de município.

Depois, usamos esses índices para obter o código da estação correspondente e adicionamos ao DataFrame original na coluna `station_code`.

Para tratar esses *NaN*, há várias abordagens.

Duas opções comuns são:

*   **Preenchimento com média por estação e mês:** calcula-se a média histórica de cada variável para aquela estação no mês correspondente, e usa-se essa média para preencher os *NaN*.
*   **Interpolação temporal:** ordena-se os dados por data para cada estação e usa-se `interpolate()` para estimar valores intermediários. O pandas possui `DataFrame.fillna()` para preenchimento e `DataFrame.interpolate()` para interpolação linear.

```python
# Interpolação temporal por estação
df_merged = df_merged.sort_values(['station_code','date'])

# Lista de colunas para interpolação
cols_to_interpolate = ['precipit_total','pressao_med','temp_orva_med','temp_max','temp_med','temp_min','umid_ar_med','umid_ar_min','vento_rajada_max','vento_rajada_med']
```

```python
# Preenchimento com média por estação e mês
for col in cols_to_interpolate:
  df_merged[col] = df_merged.groupby(['station_code', df_merged['date'].dt.month])[col].transform(lambda x: x.fillna(x.mean()))

# Preenchimento restante (se houver) com média global da coluna
for col in cols_to_interpolate:
  df_merged[col] = df_merged[col].fillna(df_merged[col].mean())
```

```python
def gerar_dicionario_datasets_multivariados_climaticos(df_grouped, df_top1_lags_climaticas):
    """
    Para cada município presente no df_top1_lags_climaticas, aplica os lags ideais nas variáveis climáticas
    e gera um DataFrame com as features defasadas + notificações. Retorna um dicionário com esses DataFrames.
    """

    import pandas as pd

    dicionario_datasets = {}

    # Obter todos os municípios únicos com lags definidos
    municipios = df_top1_lags_climaticas['municipio'].unique()

    for municipio in municipios:
        # Filtra os dados agregados semanais para o município atual
        df_mun = df_grouped[df_grouped['nome_municipio'] == municipio].copy()
        df_mun.sort_values('data_semana', inplace=True)

        # Filtra os lags ideais para esse município (um por variável climática)
        df_lags = df_top1_lags_climaticas[df_top1_lags_climaticas['municipio'] == municipio]

        # Aplica o lag (defasagem) em cada variável
        for _, row in df_lags.iterrows():
            var = row['variavel']
            lag = int(row['lag'])  # Lag já está em semanas
            nova_coluna = f"{var}_lag{lag}"
            df_mun[nova_coluna] = df_mun[var].shift(lag)

        # Define a variável target
        y = df_mun['total_notificacoes']

        # Lista as colunas defasadas que acabamos de criar
        colunas_features = [f"{row['variavel']}_lag{int(row['lag'])}" for _, row in df_lags.iterrows()]

        # Monta o DataFrame final com X e y juntos
        df_final = df_mun[colunas_features + ['data_semana', 'total_notificacoes']].dropna()

        # Salva no dicionário
        dicionario_datasets[municipio] = df_final.reset_index(drop=True)

    return dicionario_datasets

dicionario_multivariados = gerar_dicionario_datasets_multivariados_climaticos(
    df_grouped=df_grouped,
    df_top1_lags_climaticas=df_top1_lags_climaticas
)
```

```python
def gerar_dicionario_datasets_multivariados_puro(
    df_grouped: pd.DataFrame,
    df_top1_lags_climaticas: pd.DataFrame = None,
    usar_apenas_variaveis_da_tabela: bool = True,
    cols_fixas=("data_semana", "nome_municipio", "total_notificacoes")
):
    """
    Cria um dicionário {municipio: DataFrame} com as variáveis CLIMÁTICAS SEM LAG.
    - Se usar_apenas_variaveis_da_tabela=True, seleciona somente as variáveis
      que aparecem em df_top1_lags_climaticas['variavel'] (sem aplicar defasagem).
    - Caso contrário, usa todas as colunas de df_grouped exceto cols_fixas.
    """

    import pandas as pd

    # sanity
    for c in cols_fixas:
        if c not in df_grouped.columns:
            raise ValueError(f"Coluna obrigatória ausente em df_grouped: {c}")

    dicionario = {}

    # Variáveis candidatas (sem lag)
    if usar_apenas_variaveis_da_tabela:
        if df_top1_lags_climaticas is None or 'variavel' not in df_top1_lags_climaticas.columns:
            raise ValueError("Forneça df_top1_lags_climaticas com coluna 'variavel' ou use usar_apenas_variaveis_da_tabela=False.")
        variaveis_unicas = sorted(df_top1_lags_climaticas['variavel'].astype(str).unique().tolist())
    else:
        variaveis_unicas = sorted([c for c in df_grouped.columns if c not in cols_fixas])

    # Municipios a processar (pelas lags ou pelo próprio df_grouped)
    if usar_apenas_variaveis_da_tabela:
        municipios = sorted(df_top1_lags_climaticas['municipio'].astype(str).unique().tolist())
    else:
        municipios = sorted(df_grouped['nome_municipio'].astype(str).unique().tolist())

    for municipio in municipios:
        # filtra município e ordena por data
        df_mun = df_grouped[df_grouped['nome_municipio'] == municipio].copy()
        if df_mun.empty:
            continue

        # garante datetime e ordenação
        df_mun['data_semana'] = pd.to_datetime(df_mun['data_semana'])
        df_mun.sort_values('data_semana', inplace=True)

        # checa variáveis existentes neste município
        variaveis_existentes = [v for v in variaveis_unicas if v in df_mun.columns]

        if not variaveis_existentes:
            # nada a adicionar além de y + data
            df_final = df_mun[['data_semana', 'total_notificacoes']].copy()
        else:
            cols = variaveis_existentes + ['data_semana', 'total_notificacoes']
            # dropa linhas com NaN nas features escolhidas OU no alvo
            df_final = df_mun[cols].dropna(subset=variaveis_existentes + ['total_notificacoes']).reset_index(drop=True)

        dicionario[municipio] = df_final

    return dicionario

dicionario_multivariados_puro = gerar_dicionario_datasets_multivariados_puro(
    df_grouped=df_grouped,
    df_top1_lags_climaticas=df_top1_lags_climaticas,     # reutiliza a lista de variáveis, mas sem lag
    usar_apenas_variaveis_da_tabela=True                 # ou False para pegar TODAS as exógenas do df_grouped
)
```

```python
# Funções Auxiliares

def calcular_metricas(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((y_true - y_pred) / np.maximum(np.abs(y_true), 1))) * 100
    return mae, rmse, mape
```

### Modelos Lineares Generalizados (GLM) de Contagem

A predição de notificações semanais de dengue envolve dados inteiros, não negativos, com muitos zeros e possível sobredispersão (variância maior que a média). Para esse tipo de dado, modelos lineares clássicos não são adequados. Utilizamos modelos da classe GLM (Modelos Lineares Generalizados) para contagens.

---

**Regressão de Poisson**

- **Objetivo:** Modelar contagens com distribuição Poisson.
- **Link:** Logarítmico → `log(E[Y|X]) = Xβ`
- **Interpretação:** `exp(β)` indica o efeito multiplicativo da variável explicativa sobre o número esperado de casos.
- **Limite:** Pressupõe *equidispersão* (variância ≈ média). Se esse pressuposto for violado (comum), há risco de erro nas inferências.

---

**Regressão Binomial Negativa (NB)**

- **Objetivo:** Estender o modelo de Poisson para lidar com *sobredispersão*.
- **Diferencial:** Introduz um parâmetro de dispersão que permite variância maior que a média.
- **Quando usar:** Quando `var(Y) >> E(Y)` ou se o modelo de Poisson apresentar erro padrão subestimado.
- **Interpretação:** Mesma lógica do Poisson (`exp(β)`), mas com estimativas mais robustas.

---

**Modelos Inflacionados em Zero (ZIP e ZINB)**

- **Objetivo:** Modelar dados com *excesso de zeros*, além da sobredispersão.
- **Estrutura:** Duas partes:
  - Um modelo logístico que estima a chance de zero estrutural.
  - Um modelo Poisson ou NB para as contagens restantes.
- **ZIP:** Usa Poisson na parte de contagem.
- **ZINB:** Usa Binomial Negativa na parte de contagem (útil se também houver sobredispersão).
- **Interpretação:** Dois conjuntos de coeficientes – um para a contagem (efeito sobre os casos), outro para a chance de zero estrutural.

---

**Diagnóstico para Escolha do Modelo**

- **Sobredispersão:** Se a variância for muito maior que a média → prefira NB ao invés de Poisson.
- **Excesso de zeros:** Se os zeros observados forem muito mais frequentes do que o modelo NB prever → considerar ZIP ou ZINB.
- **Comparação:** AIC, BIC ou testes como Vuong ajudam a escolher entre modelos.

---

**Aplicação no Projeto**

Modelos GLM para contagem são ideais para previsão de casos de dengue, pois:
- Tratam diretamente a natureza discreta dos dados.
- Lidam com sobredispersão e excesso de zeros (muito comuns em séries epidemiológicas).
- Possuem interpretação epidemiológica clara via razão de taxas (IRR: `exp(β)`).

Começamos pelos modelos mais simples (Poisson), avaliamos os pressupostos, e evoluímos conforme necessário para NB, ZIP e ZINB.

```python
# Definição da função para realização do teste de razão de verossimilhança

# Definição da função 'lrtest'

def lrtest(modelos):
    modelo_1 = modelos[0]
    llk_1 = modelo_1.llnull
    llk_2 = modelo_1.llf

    if len(modelos)>1:
        llk_1 = modelo_1.llf
        llk_2 = modelos[1].llf
    LR_statistic = -2*(llk_1-llk_2)
    p_val = stats.chi2.sf(LR_statistic, 1) # 1 grau de liberdade

    print("Likelihood Ratio Test:")
    print(f"-2.(LL0-LLm): {round(LR_statistic, 2)}")
    print(f"p-value: {p_val:.3f}")
    print("")
    print("==================Result======================== \n")
    if p_val <= 0.05:
        print("H1: Different models, favoring the one with the highest Log-Likelihood")
    else:
        print("H0: Models with log-likelihoods that are not statistically different at 95% confidence level")

# Teste de de razão de verossimilhança para comparar as estimações dos
#'modelo_poisson' e 'modelo_bneg'

lrtest([modelo_poisson, modelo_bneg])
```

### Modelos Baseados em Árvores de Decisão e Ensemble

Nesta seção, utilizamos algoritmos baseados em árvores de decisão para modelar a série temporal multivariada de notificações de dengue.

**Por que usar Árvores de Decisão?**

Modelos de árvores são particularmente úteis por sua capacidade de capturar relações não-lineares e interações entre variáveis, sem necessidade de escalonamento dos dados. São também robustos a outliers e fáceis de interpretar.

---

**Modelos Utilizados:**

**1. Decision Tree Regressor**
- Algoritmo que particiona o espaço de entrada em regiões retangulares, ajustando uma previsão constante em cada uma.
- Vantagens:
  - Interpretação simples
  - Lida bem com variáveis categóricas ou contínuas
- Limitações:
  - Alta tendência ao overfitting (especialmente em dados ruidosos)

**2. Random Forest Regressor**
- Conjunto de árvores com amostragem aleatória e média dos resultados
- Reduz a variância do modelo

```python
# -------------------------------
# Métricas (mesmo cálculo que você usa)
# -------------------------------
def calcular_metricas(y_true, y_pred):
    mae  = np.mean(np.abs(y_true - y_pred))
    rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
    mape = np.mean(np.abs((y_true - y_pred) / np.where(y_true == 0, 1e-8, y_true))) * 100
    return mae, rmse, mape

# -------------------------------
# 1) LSTM – HISTÓRICO (AR)
# -------------------------------
def lstm_historico(df_mun):
    # seeds iguais às do seu código
    np.random.seed(42); random.seed(42); tf.random.set_seed(42)

    data = df_mun.copy()
    data['date'] = pd.to_datetime(data['data_semana'])
    data_train = data[data['date'].dt.year <= 2022]
    data_test  = data[data['date'].dt.year == 2023]

    # somente a série alvo
    notificacoes = data_train.filter(['total_notificacoes'])
    dataset = notificacoes.values

    # scaler MinMax no alvo (como no seu histórico)
    scaler = MinMaxScaler()
    scaled_data = scaler.fit_transform(dataset)

    # janelas = 52
    lookback = 52
    X_train, y_train = [], []
    for i in range(lookback, len(scaled_data)):
        X_train.append(scaled_data[i-lookback:i])
        y_train.append(scaled_data[i])
    X_train, y_train = np.array(X_train), np.array(y_train)
    X_train = X_train.reshape(X_train.shape[0], lookback, 1)

    # modelo idêntico
    model = keras.models.Sequential([
        keras.layers.LSTM(64, return_sequences=True, input_shape=(lookback, 1)),
        keras.layers.LSTM(64, return_sequences=False),
        keras.layers.Dense(128, activation='relu'),
        keras.layers.Dropout(0.2),
        keras.layers.Dense(1)
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
                  loss=tf.keras.losses.Huber(),
                  metrics=[keras.metrics.RootMeanSquaredError()])
    es = keras.callbacks.EarlyStopping(monitor='loss', patience=10, restore_best_weights=True)
    model.fit(X_train, y_train, epochs=60, batch_size=32, verbose=0, callbacks=[es])

    # teste: rolling-forecast com previsões alimentando a janela
    test_data_len = len(data_test)
    janela = scaled_data[-lookback:].copy()
    preds_scaled = []
    for _ in range(test_data_len):
        x = janela.reshape(1, lookback, 1)
        y_hat = model.predict(x, verbose=0)[0, 0]
        preds_scaled.append(y_hat)
        janela = np.vstack([janela[1:], [[y_hat]]])

    y_pred = scaler.inverse_transform(np.array(preds_scaled).reshape(-1, 1)).ravel()
    y_true = data_test['total_notificacoes'].values.astype(float)

    mae, rmse, mape = calcular_metricas(y_true, y_pred)
    return {'mae': float(mae), 'rmse': float(rmse), 'mape': float(mape)}

# -------------------------------
# 2) LSTM – CLIMÁTICO (apenas exógenas)
# -------------------------------
def lstm_climatico(df_mun):
    np.random.seed(42); random.seed(42); tf.random.set_seed(42)

    data = df_mun.copy()
    data['date'] = pd.to_datetime(data['data_semana'])
    data_train = data[data['date'].dt.year <= 2022]
    data_test  = data[data['date'].dt.year == 2023]

    # exógenas = tudo exceto data/date/alvo
    feature_cols = [c for c in data.columns if c not in ['data_semana', 'date', 'total_notificacoes']]
    X_train_df = data_train[feature_cols].copy()
    y_train_df = data_train[['total_notificacoes']].copy()

    # scalers (Robust para X e y, como no seu código)
    scaler_X = RobustScaler()
    scaler_y = RobustScaler()
    X_scaled = scaler_X.fit_transform(X_train_df.values)
    y_scaled = scaler_y.fit_transform(y_train_df.values)

    # lookback = 1 (conforme seu script climático)
    lookback = 1
    X_train, y_train = [], []
    for i in range(lookback, len(X_scaled)):
        X_train.append(X_scaled[i-lookback:i, :])
        y_train.append(y_scaled[i, 0])
    X_train, y_train = np.array(X_train), np.array(y_train)
    n_features = X_train.shape[2]

    model = keras.models.Sequential([
        keras.layers.LSTM(64, return_sequences=True, input_shape=(lookback, n_features)),
        keras.layers.LSTM(64, return_sequences=False),
        keras.layers.Dense(128, activation='relu'),
        keras.layers.Dropout(0.2),
        keras.layers.Dense(1)
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
                  loss=tf.keras.losses.Huber(),
                  metrics=[keras.metrics.RootMeanSquaredError()])
    es = keras.callbacks.EarlyStopping(monitor='loss', patience=10, restore_best_weights=True)
    model.fit(X_train, y_train, epochs=60, batch_size=32, verbose=0, callbacks=[es])

    # teste one-step com EXÓGENAS reais de 2023
    test_data_len = len(data_test)
    X_test_block = scaler_X.transform(data_test[feature_cols].values)

    # bloco contínuo: últimas 'lookback' do treino + 2023
    X_block = np.vstack([X_scaled[-lookback:, :], X_test_block])

    # janelas para 2023
    X_test_windows = []
    for i in range(lookback, lookback + test_data_len):
        X_test_windows.append(X_block[i-lookback:i, :])
    X_test = np.array(X_test_windows)

    y_pred_scaled = model.predict(X_test, verbose=0)
    y_pred = scaler_y.inverse_transform(y_pred_scaled).ravel()

    y_true = data_test['total_notificacoes'].values.astype(float)
    mae, rmse, mape = calcular_metricas(y_true, y_pred)
    return {'mae': float(mae), 'rmse': float(rmse), 'mape': float(mape)}

# -------------------------------
# 3) LSTM – HISTÓRICO + CLIMA (multivariado)
# -------------------------------
def lstm_hist_clima(df_mun):
    np.random.seed(42); random.seed(42); tf.random.set_seed(42)

    data = df_mun.copy()
    data['date'] = pd.to_datetime(data['data_semana'])
    data_train = data[data['date'].dt.year <= 2022]
    data_test  = data[data['date'].dt.year == 2023]

    # features: todas exceto colunas de data
    feature_cols = [c for c in data.columns if c not in ['data_semana', 'date']]
    exo_cols = [c for c in feature_cols if c != 'total_notificacoes']
    ar_col  = ['total_notificacoes']

    X_train_df = data_train[feature_cols].copy()
    y_train_df = data_train[ar_col].copy()

    # scalers (MinMax nas exógenas; Robust para AR e y), conforme seu script
    scaler_exo = MinMaxScaler()
    scaler_ar  = RobustScaler()

    X_exo_scaled = scaler_exo.fit_transform(X_train_df[exo_cols].values)
    X_ar_scaled  = scaler_ar.fit_transform(X_train_df[ar_col].values)

    # ordem: exógenas + AR
    feature_cols_ordered = exo_cols + ar_col
    X_train_scaled = np.column_stack([X_exo_scaled, X_ar_scaled])

    # y escalado com o MESMO scaler_ar
    y_train_scaled = scaler_ar.transform(y_train_df.values).ravel()

    # janelas = 52
    lookback = 52
    def build_windows(X_scaled, y_scaled, window):
        X_seq, Y_seq = [], []
        for i in range(window, len(X_scaled)):
            X_seq.append(X_scaled[i-window:i, :])
            Y_seq.append(y_scaled[i])
        return np.array(X_seq), np.array(Y_seq)

    X_train, y_train = build_windows(X_train_scaled, y_train_scaled, lookback)
    n_features = X_train.shape[2]

    model = keras.models.Sequential([
        keras.layers.LSTM(64, return_sequences=True, input_shape=(lookback, n_features)),
        keras.layers.LSTM(64, return_sequences=False),
        keras.layers.Dense(128, activation='relu'),
        keras.layers.Dropout(0.2),
        keras.layers.Dense(1)
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
                  loss=tf.keras.losses.Huber(),
                  metrics=[keras.metrics.RootMeanSquaredError()])
    es = keras.callbacks.EarlyStopping(monitor='loss', patience=10, restore_best_weights=True)
    model.fit(X_train, y_train, epochs=60, batch_size=32, verbose=0, callbacks=[es])

    # teste: exógenas reais + AR autoregressivo via previsão
    test_data_len = len(data_test)
    X_test_df = data_test[feature_cols_ordered].copy()

    X_test_exo_scaled = scaler_exo.transform(X_test_df[exo_cols].values)
    X_test_ar_scaled_true = scaler_ar.transform(X_test_df[ar_col].values)  # só referência (não alimenta)
    X_test_scaled_base = np.column_stack([X_test_exo_scaled, X_test_ar_scaled_true])

    janela = X_train_scaled[-lookback:, :].copy()
    idx_y = feature_cols_ordered.index('total_notificacoes')

    preds_scaled = []
    for i in range(test_data_len):
        x = janela.reshape(1, lookback, n_features)
        y_hat_scaled = model.predict(x, verbose=0)[0, 0]
        preds_scaled.append(y_hat_scaled)

        proxima_linha = X_test_scaled_base[i, :].copy()
        proxima_linha[idx_y] = y_hat_scaled
        janela = np.vstack([janela[1:, :], proxima_linha.reshape(1, -1)])

    y_pred = scaler_ar.inverse_transform(np.array(preds_scaled).reshape(-1, 1)).ravel()
    y_true = data_test['total_notificacoes'].values.astype(float)

    mae, rmse, mape = calcular_metricas(y_true, y_pred)
    return {'mae': float(mae), 'rmse': float(rmse), 'mape': float(mape)}

# -------------------------------
# Função principal: roda tudo e devolve a tabela final
# -------------------------------
def gerar_tabela_resultados_lstm(dicionario_multivariados, municipios=None):
    """
    dicionario_multivariados: dict[str, pd.DataFrame]
    municipios: lista opcional de nomes para filtrar; se None, usa todos.
    """
    resultados = []
    lista_muns = municipios if municipios is not None else list(dicionario_multivariados.keys())

    for mun in lista_muns:
        df_mun = dicionario_multivariados[mun].copy()

        # HISTÓRICO
        res_hist = lstm_historico(df_mun)
        resultados.append({'Município': mun, 'Modelo': 'LSTM', 'Tipo': 'Histórico',
                           'MAE': res_hist['mae'], 'RMSE': res_hist['rmse'], 'MAPE': res_hist['mape']})

        # CLIMÁTICO
        res_clima = lstm_climatico(df_mun)
        resultados.append({'Município': mun, 'Modelo': 'LSTM', 'Tipo': 'Clima',
                           'MAE': res_clima['mae'], 'RMSE': res_clima['rmse'], 'MAPE': res_clima['mape']})

        # HISTÓRICO + CLIMA
        res_hc = lstm_hist_clima(df_mun)
        resultados.append({'Município': mun, 'Modelo': 'LSTM', 'Tipo': 'Histórico+Clima',
                           'MAE': res_hc['mae'], 'RMSE': res_hc['rmse'], 'MAPE': res_hc['mape']})

    df_resultados_lstm = pd.DataFrame(resultados, columns=['Município','Modelo','Tipo','MAE','RMSE','MAPE'])
    return df_resultados_lstm

df_resultados_lstm = gerar_tabela_resultados_lstm(dicionario_multivariados)
df_resultados_lstm.sort_values(['Município','RMSE'])
```
